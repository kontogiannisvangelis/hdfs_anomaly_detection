```python
from datasets import load_from_disk

ds = load_from_disk("/home/ekontogi/Diploma-Thesis/dataset/tokenized_dataset")
dataset = ds['train']
```

```python
for i in range(3):
    row = dataset[i]

    values = " | ".join(
        f"{col}: {row.get(col)}"
        for col in dataset.column_names
        if col in row
    )

    print(f"{i+1}. {values}")
```
